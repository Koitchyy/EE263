```julia
include(joinpath(@__DIR__, "..", "readclassjson.jl"))
using LinearAlgebra
using Plots
```

```julia
data = readclassjson("lena_data.json")
X = data["X"]

U, S, V = svd(X)

idx = 200
p_a = scatter(idx:size(X, 1), S[idx:size(X, 1)], xlabel="i", 
              ylabel="σ_i", legend=false, title="σ$idx(X) ~ σn(X)")
```

### Save Image Storage Space

```julia
r = 75
U_tilda = U[:,1:r]
S_tilda = diagm(S[1:r])
V_tilda = V[:,1:r]

X_tilda = U_tilda * S_tilda * V_tilda'

rel_error = norm(X[:] - X_tilda[:]) / norm(X[:])

println("The relative Frobenius norm error is: $rel_error")
```
