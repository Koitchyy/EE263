```julia
using Distributions
using Plots

dist_xy = Uniform(-1, 1)
x_samples = rand(dist_xy, 10000)
y_samples = rand(dist_xy, 10000)
plot_z = scatter(x_samples, y_samples, xlabel="x", ylabel="y", 
                 title="10000 samples of z", legend=false, aspect_ratio = :equal)

dist_pq = Normal(0, 1)
p_samples = rand(dist_pq, 10000)
q_samples = rand(dist_pq, 10000)
plot_w = scatter(p_samples, q_samples, xlabel="p", ylabel="q", 
                 title="10000 samples of w", legend=false, aspect_ratio = :equal)

display(plot_z)
display(plot_w)
```
